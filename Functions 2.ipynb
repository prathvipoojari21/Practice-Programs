```python
#Program that accepts three numbers from user and display its addition using function
x=int(input("Enter x"))
y=int(input("Enter y"))
z=int(input("Enter z"))

def cal_sum(x,y,z):
    sum=x+y+z
    return sum

s=cal_sum(x,y,z)
print(s)
```

```python
#Program to receive three integers from keyboard and get their sum and product calculated through a user defined function cal_sum_prod()
def cal_sum_prod(x,y,z):
    ss=x+y+z
    pp=x*y*z
    return ss,pp
a=int(input("Enter a: "))
b=int(input("Enter b: "))
c=int(input("Enter c: "))
s,p=cal_sum_prod(a,b,c)
print(s,p)
```
