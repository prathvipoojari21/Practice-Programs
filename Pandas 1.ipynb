```python
#Creating a series from a list
import pandas as pd
a=[1,7,2]
myvar=pd.Series(a)
print(myvar)
```

```python
#Create Labels
import pandas as pd
a=[1,7,2]
myvar=pd.Series(a,index=["x","y","z"])
print(myvar)
```

```python
import pandas as pd
s1=pd.Series([10,20,30],index=['a','b','c'],name='Example Series')
print(s1)
```

```python
#Creating a Series from a dictionary
import pandas as pd
s2=pd.Series({'x':1,'y':2,'z':3})
print(s2)
```

```python
import pandas as pd
calories={"day1":420,"day2":380,"day3":390}
myvar=pd.Series(calories)
print(myvar)
```
