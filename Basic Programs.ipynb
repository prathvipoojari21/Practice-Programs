```python
#Simple Calculator
num1=float(input("Enter first number:"))
num2=float(input("Enter second number:"))
print("\nChoose an operation: , 1. Add , 2. Subtract , 3. Multiply, 4. Divide")
choice=int(input("Enter your choice:"))
if choice==1:
 print("The addition of the numbers are: ", num1+num2)
elif choice==2:
 print("The subtraction of the numbers are: ", num1-num2)
elif choice==3:
 print("The multiplication of the numbers are: ", num1*num2)
elif choice==4:
    if num2!=0:
     print("The division of the numbers are: ", num1/num2)
    else:
     print("Cannot divide")
else:
 print("Invalid choice")
```
