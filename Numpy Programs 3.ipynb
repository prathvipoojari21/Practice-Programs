```python
#Creating a 3x3 NumPy array of all True
import numpy as np
array_true=np.full((3,3),True)
print(array_true)
```

```python
#Creating a 1D array of 10 evenly spaced values between 5 and 50
import numpy as np
spaced_array=np.linspace(5,50,10)
print(spaced_array)
```

```python
#Converting python list to NumPy array
import numpy as np
python_list=[10,20,30,40,50]
numpy_array=np.array(python_list)
print(numpy_array)
```

```python
#Reversing a 1D NumPy array
import numpy as np
ori_array=np.array([1,2,3,4,5])
rev_array=ori_array[::-1]
print(rev_array)
```

```python
#Creating a 3x3 identity matrix
import numpy as np
identity_matrix=np.eye(3)
print(identity_matrix)
```
