```python
#Data Cleaning
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
#use dropna() to remove rows with any missing values
df_cleaned=df.dropna()
print("Cleansed Data")
print(df_cleaned)
```

```python
#Fill missing values
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()
#filling NaN va;ues with 0
df.fillna(0,inplace=True)
print("Data after filling missing values")
print(df)
```

```python
#Use aggregate functions to fill missing values
import pandas as pd
data={
    'A':[1,2,3,None,5],
    'B':[None,2,3,4,5],
    'C':[1,2,None,None,5]
}
df=pd.DataFrame(data)
print("Original Data:\n",df)
print()

#filling NaN values with the mean of each column
df.fillna(df.mean(),inplace=True)
print("Data after filling missing values:\n",df)
```

```python
#Handle dupilcate values
import pandas as pd
#sample data
data={
    'A':[1,2,2,3,3,4],
    'B':[5,6,6,7,8,8]
}
df=pd.DataFrame(data)
print("Original DataFrame:\n",df.to_string(index=False))
#detect duplicates
print("Duplicate Rows:\n",df[df.duplicated()].to_string(index=False))
#remove duplicates based on column 'A'
df.drop_duplicates(subset=['A'],keep='first',inplace=True)
print("DataFrame after removing duplicates based column 'A':\n",df.to_string(index=False))
```

```python
#Rename column names to meaningful names
import pandas as pd
#sample data
data={
    'A':[25,30,35],
    'B':['John','Doe','Smith'],
    'C':[50000,60000,70000]
}
df=pd.DataFrame(data)
#rename columns
df.rename(columns={'A':'Age','B':'Name','C':'Salary'},inplace=True)
print(df.to_string(index=False))
```
