```python
#Program to sort keys in a dictionary
my_dict={
    "name":"ABC",
    "panel":"B",
    "rollno":34,
    "marks":[65,87,67,94]}
sorted_dict=sorted(my_dict)
print("Sorted Keys: ",sorted_dict)
```

```python
#Program to create a dictionary of 5 students with their marks and display the student with highest marks
students={
    "Amit":88,
    "Tina":90,
    "Tanya":77,
    "Roshi":79,
    "Aarav":95}
highest_student=max(students,key=students.get)
print("Student with highest marks: ",highest_student)
print("Highest Marks: ",students[highest_student])
```

```python
#Program to count the frequency of each element in a list using dictionary
numbers=[10,20,30,30,10,50,60]
frequency={}
for num in numbers:
    if num in frequency:
        frequency[num]=frequency[num]+1
    else:
        frequency[num]=1
print("Frequency of elements: ",frequency)
```

```python
#Program to sort a dictionary by its values
students={
    "Amit":88,
    "Tina":90,
    "Tanya":77,
    "Roshi":79,
    "Aarav":95}
sorted_dict=dict(sorted(students.items(),key=lambda item: item[1]))
print("Dictionary sorted by values: ")
print(sorted_dict)
#For descending order
sorted_dict_new=dict(sorted(students.items(),key=lambda item: item[1], reverse=True))
print("Descending order:")
print(sorted_dict_new)
```
