```python
numbers=[]
for i in range(10):
    num=int(input("Enter a number"))
    numbers.append(num)
print("\Original List: ",numbers)

#Ascending order using sorted()
ascending=sorted(numbers)
print("\nAscending order: ",ascending)

#Descending order
numbers.sort(reverse=True)
print("\nDescending order: ", numbers)

#Length of the list
print("\nThe length of the list is: ",len(numbers))
```

```python
#Accepting 2 lists and merging them
list1=[]
list2=[]
for i in range(5):
    num_1=int(input("Enter a number"))
    list1.append(num_1)
print("\nList 1: ",list1)
for j in range(5):
    num_2=int(input("Enter a number"))
    list2.append(num_2)
print("\nList 2: ",list2)
merged_list=list1+list2
print("\nMerged List: ",merged_list)
```

```python
phrase=input("Enter a phrase")
words=phrase.split()
acronym=""

for word in words:
    acronym=acronym+word[0]

print("Acronym: ",acronym.upper())
```

```python
#Program to print abbreviation of a month, given its number
months=["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
m_n=int(input("Enter month number(1-12)): "))
if m_n>=1 and m_n<=12:
    print("Month Abbreviation: ",months[m_n-1])
else:
    print("Inavlid month number")
```
